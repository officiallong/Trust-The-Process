# Exercise 3: Self-Reflection Prompt for Improving Output

**Student:** Long Nguyen  
**Tool:** Google Colab, Python, Google Gemini API  
**Goal:** Use a self-reflection prompt to critique an AI-generated summary and create an improved version based on specific evaluation criteria.

**## 1. Gemini API Setup**

```python
!pip install -q -U google-genai
```

```python
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

print("Gemini client connected successfully.")
```

```python
def ask_ai(prompt):
    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )
    return response.text
```

## 2. Generate the Original Summary**

```python
original_prompt = """
You are a business writing assistant.

Summarize the following information about artificial intelligence in
customer service for a college business student.

Information:
Artificial intelligence is increasingly used in customer service to
automate routine interactions and help customers receive support more
quickly. AI-powered chatbots can answer common questions, provide
basic troubleshooting, and route more complex issues to human agents.
AI can also help businesses analyze customer interactions and identify
common problems. However, businesses should consider challenges such
as inaccurate responses, customer privacy, data security, and knowing
when human assistance is necessary.

Write one short paragraph summarizing the information.
"""

original_summary = ask_ai(original_prompt)

print("ORIGINAL SUMMARY")
print(original_summary)
```

## 3.Self-Reflection and Critique**

```python
reflection_prompt = f"""
You are a business writing editor reviewing an AI-generated summary.

Here is the original summary:

{original_summary}

Critique the summary using each of the following criteria:

1. Accuracy:
Does the summary accurately represent the information provided without
adding unsupported claims?

2. Clarity:
Is the summary easy to understand, or is any wording unnecessarily
complex or vague?

3. Audience:
Is the language appropriate for a college business student?

4. Length:
Is the summary concise while still including the most important points?

5. Formatting:
Is the one-paragraph format appropriate and easy to read?

For each criterion:
- Identify what the summary does well.
- Identify a specific improvement if one is needed.
- Do not rewrite the summary yet.

Then provide 2-3 specific recommendations that should be used when
creating the revised version.

Use exactly this format:

ACCURACY:
[critique]

CLARITY:
[critique]

AUDIENCE:
[critique]

LENGTH:
[critique]

FORMATTING:
[critique]

RECOMMENDATIONS:
[2-3 specific recommendations]
"""

reflection = ask_ai(reflection_prompt)

print("SELF-REFLECTION AND CRITIQUE")
print(reflection)
```

**## 4. Generate the Improved Summary**

```python
improvement_prompt = f"""
You are a business writing assistant revising an AI-generated summary
based on a self-reflection critique.

ORIGINAL SUMMARY:
{original_summary}

SELF-REFLECTION AND CRITIQUE:
{reflection}

Revise the original summary using the feedback from the critique.

Requirements:
- Preserve the factual meaning of the original source information.
- Do not introduce unsupported facts or claims.
- Improve clarity by simplifying unnecessarily complex wording.
- Use language appropriate for a college business student.
- Keep the revised summary concise at approximately 70-100 words.
- Use one paragraph with about 3 sentences.
- Clearly explain the role of AI in customer service.
- Include both benefits and potential challenges.
- Make human escalation easy to understand.
- Only include business impacts that are directly supported by the
  original information.

Return only the revised summary with no additional explanation.
"""

improved_summary = ask_ai(improvement_prompt)

print("IMPROVED SUMMARY")
print(improved_summary)
```

### **5. Before and After Comparison**

```python
print("BEFORE - ORIGINAL SUMMARY")
print(original_summary)

print("\nAFTER - IMPROVED SUMMARY")
print(improved_summary)
```

## 6. Testing and Reflection

The original summary was accurate and concise, but the self-reflection identified several areas that could be improved. In particular, some of the wording was overly complex, the second sentence contained several ideas at once, and the explanation of when human assistance is needed could be clearer.

I used five criteria to evaluate the summary: accuracy, clarity, audience, length, and formatting. Based on the critique, I asked the AI to revise its original response while preserving the factual meaning and avoiding unsupported claims. The improved version separates the main ideas more clearly, uses three sentences instead of two, and explains the handoff to human agents more directly.

This exercise showed how self-reflection can be used as an additional step in prompt engineering. Instead of accepting the first AI response, I had the model evaluate its own work using specific criteria and then used that feedback to generate a revised response.
